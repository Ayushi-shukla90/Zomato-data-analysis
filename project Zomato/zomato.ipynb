```python
import pandas as pd
from sqlalchemy import create_engine

df = pd.read_csv(r"d:\zomato_1000_rows.csv")
df.info()
df.head()
df.columns
df.drop(columns=["url"], inplace=True)
df.columns
df.drop(columns=["menu_item"], inplace=True)
df.head(20)
# Remove the denominator from the rate column, e.g. "4.5/5" -> "4.5"
df['rate'] = df['rate'].astype(str).str.split('/').str[0].str.strip()
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
df.head(20)
df.drop(columns=["reviews_list"], inplace=True)
df.head(20)

# Clean column names
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Rename column
df.rename(
    columns={'approx_cost(for_two_people)': 'approx_cost'},
    inplace=True
)
df.head(20)
df.rename(
    columns={'listed_in(type)': 'listed_in_type'},
    inplace=True
)
df.rename(
    columns={'listed_in(city)': 'listed_in_city'},
    inplace=True
)
df.rename(
    columns={'name': 'restaurant'},
    inplace=True
)
df.head(20)
df = df.dropna()
print("removed null values")
# MySQL Connection
username = "root"
password = "netware"
host = "127.0.0.1"
port = 3306
database = "zomato_data"


# Connection string
connection_string = (
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

# Create engine
engine = create_engine(connection_string)


df.to_sql(
    name='zomato_data',
    con=engine,
    if_exists='replace',
    index=False
)

print("Data uploaded successfully to MySQL")

# export csv file
df.to_csv(r"d:\cleaned_zomato_data.csv", index=False)
print("Cleaned data exported to CSV successfully")
```
